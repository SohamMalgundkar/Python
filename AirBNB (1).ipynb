```python
df=pd.read_csv("compressed_data.csv")
```

# Changing type of last review Column to Date

```python
df['last review']=pd.to_datetime(df['last review'],errors='coerce')  #coerce means When we change type to date it will change the blank value to 'NAT'format i.e is Not A Thing
```

```python
df.info()
```

# **Handling Missing Data**

```python
df.fillna({'reviews per month':0 ,'last review':df["last review"].min()},inplace=True)
```

```python
df.dropna(subset=['NAME','host name'],inplace=True)
```

```python
df.isnull().sum()
```

```python
df.drop(columns=["house_rules","license"])
```

```python
df['price']=df['price'].replace('[\$,]','',regex=True).astype(float)
df['service fee']=df['service fee'].replace('[\$,]','',regex=True).astype(float)
```

```python
df.head()
```

# Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

# Discriptive Stats

```python
df.describe()
```

# Visualisation

# Distribution of listing Price


```python
sns.histplot(df['price'], bins=50, kde=True,color='teal')
plt.title('Distribution of Price')
plt.ylabel('Frequency')
plt.figure(figsize=(10,6))
plt.show()
```

# Room Type Distribution

```python
sns.countplot(data=df,x='room type',color='cyan')
plt.title('Room Type Distribution')
plt.ylabel('Count')
plt.show()
```

# Listing Across Different Neighbourhood Groups

```python
sns.countplot(data=df,y='neighbourhood group',order=df['neighbourhood group'].value_counts().index)
plt.title('Distribution in neighbourhood')
plt.figure(figsize=(12,8))
```

# Room type vs price

```python
sns.boxplot(data=df,x='room type', y='price',hue='room type',palette='Set1')
plt.title('Room type VS Price')
plt.show()
```

# Estimated Revenue

```python
df['estimated_revenue'] = df['price'] * (365 - df['availability 365'])
```

```python
sns.barplot(data=df,x='neighbourhood group',y='estimated_revenue')
```
